```python
import pandas as pd
import numpy as np
from collections import defaultdict, deque
from scipy.optimize import linear_sum_assignment
from tqdm.contrib.concurrent import process_map
from tqdm import tqdm
import networkx as nx
import random
import heapq
import os
```

```python
# load all du lab embryo data in 'data/more_embryo_du_lab/*.txt'

for file in os.listdir("data/more_embryo_du_lab"):
    if file.endswith(".txt"):
        track_df = pd.read_csv(os.path.join("data/more_embryo_du_lab", file), sep="\t")
        max_time = track_df.time.max()
        num_cells = len(track_df.cell_name.unique())
        print(f"{file}: max_time={max_time}, num_cells={num_cells}")
```
