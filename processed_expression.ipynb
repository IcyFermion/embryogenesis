```python
lineage_names = set()
for i in exp_data['lineage'].unique():
    if "/" in i:
        for j in i.split("/"):
            lineage_names.add(j)
    else:
        lineage_names.add(i)
```

```python
lineage_names = sorted(list(lineage_names))
print(lineage_names)
```

```python
# save lineage_names to a file
with open("lineage_names_with_exp.txt", "w") as f:
    for name in lineage_names:
        f.write(name + "\n") 
```
