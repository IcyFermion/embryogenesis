```python
# format lineage names to match tree lineage names then find the indices of the tree lineage names
# in the CPM lineage dataset
formated_names = [name.decode('utf-8') if isinstance(name, bytes) else name for name in f['CPM/lineages']]
tree_lineage_names = pd.read_csv('data/tree_lineage_names.txt', header=None, names=['lineage']).values.flatten().tolist()
tree_lineage_names = list(set(tree_lineage_names).intersection(set(formated_names))) 
tree_indices = [formated_names.index(name) for name in tree_lineage_names if name in formated_names]

# since indices are not in order, sort them and sort the tree lineage names accordingly, then save the new order of names into a file
sorted_tree_indices = sorted(tree_indices)
sorted_tree_lineage_names = [formated_names[i] for i in sorted_tree_indices]
# Save the sorted lineage names to a file
with open('data/sorted_tree_lineage_names.txt', 'w') as f_out:
    for name in sorted_tree_lineage_names:
        f_out.write(f"{name}\n")
```

```python
# format gene names to match transcription factor names then find the indices of the transcription factor names
# in the CPM gene dataset
# and save the sorted transcription factor names to a file

formated_gene_names = [name.decode('utf-8') if isinstance(name, bytes) else name for name in f['CPM/genes']]
tf_names = pd.read_csv('data/dropbox_valid_tf_names.txt', header=None, names=['tf']).values.flatten().tolist()
tf_names = list(set(tf_names).intersection(set(formated_gene_names))) 
tf_indices = [formated_gene_names.index(name) for name in tf_names if name in formated_gene_names]

sorted_tf_indices = sorted(tf_indices)
sorted_tf_names = [formated_gene_names[i] for i in sorted_tf_indices]
```

```python
print("Calculating tf statistics across the dataset (tree lineages only)...")

n_genes = len(sorted_tf_names)
n_timepoints = f['CPM/data'].shape[0]
n_lineages = len(sorted_tree_indices)

# Initialize arrays to store statistics
gene_means = np.zeros(n_genes)
gene_vars = np.zeros(n_genes)
gene_nonzero_counts = np.zeros(n_genes)

chunk_size = 10  # Process 10 timepoints at a time

print(f"Processing {n_timepoints} timepoints in chunks of {chunk_size}...")

for t_start in range(0, n_timepoints, chunk_size):
    t_end = min(t_start + chunk_size, n_timepoints)
    
    # Load chunk for tree lineages only and zero-fill NaNs
    chunk = f['CPM/data'][t_start:t_end, sorted_tf_indices, :]
    chunk = chunk[:, :, sorted_tree_indices]
    chunk = np.nan_to_num(chunk, nan=0.0)
    
    # Calculate statistics for this chunk
    chunk_mean = chunk.mean(axis=(0, 2))
    print(chunk_mean.sum()) # Debugging line to check the mean values
    gene_means += chunk_mean * (t_end - t_start)
    
    chunk_sq_mean = (chunk**2).mean(axis=(0, 2))
    gene_vars += chunk_sq_mean * (t_end - t_start)
    
    gene_nonzero_counts += (chunk > 0).sum(axis=(0, 2))
    
    if (t_start + chunk_size) % 50 == 0:
        print(f"  Processed timepoints {t_start}-{t_end-1}")

gene_means /= n_timepoints
gene_vars = gene_vars / n_timepoints - gene_means**2

total_values_per_gene = n_timepoints * n_lineages
gene_detection_rate = gene_nonzero_counts / total_values_per_gene

print("\nGene statistics summary:")
print(f"  Mean expression range: {gene_means.min():.3f} to {gene_means.max():.3f}")
print(f"  Variance range: {gene_vars.min():.3f} to {gene_vars.max():.3f}")
print(f"  Detection rates: {gene_detection_rate.min():.1%} to {gene_detection_rate.max():.1%}")
```

```python
# 1. Log transformation (log1p = log(1 + x))
print("\nApplying log transformation...")
log_expression = np.log1p(avg_expression)

# 2. Normalize each lineage (cell type) to same total counts
print("Normalizing lineages...")
# Calculate normalization factors
lineage_sums = log_expression.sum(axis=0)
# Avoid division by zero
lineage_sums[lineage_sums == 0] = 1.0
# Target sum (median of all lineage sums)
target_sum = np.median(lineage_sums[lineage_sums > 0])
# Normalize
norm_expression = log_expression * (target_sum / lineage_sums)

# 3. Gene-wise scaling (z-score normalization)
print("Scaling genes...")
gene_means = norm_expression.mean(axis=1, keepdims=True)
gene_stds = norm_expression.std(axis=1, keepdims=True)
# Avoid division by zero
gene_stds[gene_stds == 0] = 1.0
scaled_expression = (norm_expression - gene_means) / gene_stds

print("\nPreprocessing complete!")
print(f"Final matrix shape: {scaled_expression.shape}")
print(f"Value range: {scaled_expression.min():.3f} to {scaled_expression.max():.3f}")
```

```python
print("Calculating gene statistics across the dataset (tree lineages only)...")

n_genes = f['CPM/data'].shape[1]
n_timepoints = f['CPM/data'].shape[0]
n_lineages = len(sorted_tree_indices)

# Initialize arrays to store statistics
gene_means = np.zeros(n_genes)
gene_vars = np.zeros(n_genes)
gene_nonzero_counts = np.zeros(n_genes)

chunk_size = 10  # Process 10 timepoints at a time

print(f"Processing {n_timepoints} timepoints in chunks of {chunk_size}...")

for t_start in range(0, n_timepoints, chunk_size):
    t_end = min(t_start + chunk_size, n_timepoints)
    
    # Load chunk for tree lineages only and zero-fill NaNs
    chunk = f['CPM/data'][t_start:t_end, :, sorted_tree_indices]
    chunk = np.nan_to_num(chunk, nan=0.0)
    
    # Calculate statistics for this chunk
    chunk_mean = chunk.mean(axis=(0, 2))
    gene_means += chunk_mean * (t_end - t_start)
    
    chunk_sq_mean = (chunk**2).mean(axis=(0, 2))
    gene_vars += chunk_sq_mean * (t_end - t_start)
    
    gene_nonzero_counts += (chunk > 0).sum(axis=(0, 2))
    
    if (t_start + chunk_size) % 50 == 0:
        print(f"  Processed timepoints {t_start}-{t_end-1}")

gene_means /= n_timepoints
gene_vars = gene_vars / n_timepoints - gene_means**2

total_values_per_gene = n_timepoints * n_lineages
gene_detection_rate = gene_nonzero_counts / total_values_per_gene

print("\nGene statistics summary:")
print(f"  Mean expression range: {gene_means.min():.3f} to {gene_means.max():.3f}")
print(f"  Variance range: {gene_vars.min():.3f} to {gene_vars.max():.3f}")
print(f"  Detection rates: {gene_detection_rate.min():.1%} to {gene_detection_rate.max():.1%}")
```

```python
# Get basic distribution info
print("\nGene expression distribution:")
print(f"  Genes with mean > 0: {(gene_means > 0).sum()} ({(gene_means > 0).sum()/n_genes*100:.1f}%)")
print(f"  Genes with mean > 0.1: {(gene_means > 0.1).sum()} ({(gene_means > 0.1).sum()/n_genes*100:.1f}%)")
print(f"  Genes with mean > 1.0: {(gene_means > 1.0).sum()} ({(gene_means > 1.0).sum()/n_genes*100:.1f}%)")

# Show percentiles
percentiles = [0, 25, 50, 75, 90, 95, 99, 100]
print("\nMean expression percentiles:")
for p in percentiles:
    val = np.percentile(gene_means, p)
    print(f"  {p}th percentile: {val:.3f}")

print("\nExpression variance percentiles:")
for p in percentiles:
    val = np.percentile(gene_vars, p)
    print(f"  {p}th percentile: {val:.3f}")
    
print("\nGene detection rate percentiles:")
for p in percentiles:
    val = np.percentile(gene_detection_rate, p)
    print(f"  {p}th percentile: {val:.5f}")
```

```python
# Approach 1: Highly Variable Genes (HVG) selection
# This is standard for scRNA-seq clustering

# First apply minimal quality filters
basic_filter = (gene_means > 0.01) & (gene_detection_rate > 0.001)  # Very permissive
print(f"Basic filter keeps: {basic_filter.sum()} genes")

# Then select top N highly variable genes
# Calculate coefficient of variation (CV) for genes passing basic filter
cv = np.sqrt(gene_vars[basic_filter]) / (gene_means[basic_filter] + 1e-6)
n_hvg = 2000  # Common choice for UMAP

# Get indices of top variable genes
basic_idx = np.where(basic_filter)[0]
top_cv_idx = np.argsort(cv)[-n_hvg:]
kept_genes_idx = basic_idx[top_cv_idx]

print(f"\nSelected {len(kept_genes_idx)} highly variable genes")
print(f"Mean range: {gene_means[kept_genes_idx].min():.3f} - {gene_means[kept_genes_idx].max():.3f}")
print(f"Detection range: {gene_detection_rate[kept_genes_idx].min():.2%} - {gene_detection_rate[kept_genes_idx].max():.2%}")
```

```python
# Approach 2: Focus on reliably detected genes
# Better for very sparse data like yours

# Prioritize detection rate over mean expression
detection_weight = 0.7
expression_weight = 0.3

# Create combined score
gene_scores = (detection_weight * gene_detection_rate / gene_detection_rate.max() + 
               expression_weight * gene_means / gene_means.max())

# Select top N genes by score
n_genes = 3000
top_genes_idx = np.argsort(gene_scores)[-n_genes:]

print(f"\nSelected {n_genes} genes by combined score")
print(f"Detection rates: {gene_detection_rate[top_genes_idx].min():.2%} - {gene_detection_rate[top_genes_idx].max():.2%}")
```

```python
# 1. Log transformation (log1p = log(1 + x))
print("\nApplying log transformation...")
log_expression = np.log1p(avg_expression)

# 2. Normalize each lineage (cell type) to same total counts
print("Normalizing lineages...")
# Calculate normalization factors
lineage_sums = log_expression.sum(axis=0)
# Avoid division by zero
lineage_sums[lineage_sums == 0] = 1.0
# Target sum (median of all lineage sums)
target_sum = np.median(lineage_sums[lineage_sums > 0])
# Normalize
norm_expression = log_expression * (target_sum / lineage_sums)

# 3. Gene-wise scaling (z-score normalization)
print("Scaling genes...")
gene_means = norm_expression.mean(axis=1, keepdims=True)
gene_stds = norm_expression.std(axis=1, keepdims=True)
# Avoid division by zero
gene_stds[gene_stds == 0] = 1.0
scaled_expression = (norm_expression - gene_means) / gene_stds

print("\nPreprocessing complete!")
print(f"Final matrix shape: {scaled_expression.shape}")
print(f"Value range: {scaled_expression.min():.3f} to {scaled_expression.max():.3f}")
```

```python
# Transpose to get lineages x genes format
expression_matrix = scaled_expression.T  # Shape: (n_lineages, n_genes)
print(f"\nFinal expression matrix: {expression_matrix.shape}")
print("Ready for distance calculations and UMAP!")

# Save for later use
np.save('processed_expression_matrix.npy', expression_matrix)
np.save('lineage_names.npy', f['CPM/lineages'][:])
np.save('selected_gene_names.npy', selected_gene_names)
print("\nSaved processed data files")
```
